```python
# Count NaNs per column
nan_counts = df.isna().sum()
print(nan_counts)
```

```python
# Total listings and hosts
total_listings = df.shape[0]
print(f"Number of total listings in Melbourne: {total_listings}")
```

```python
# Total listings and hosts
total_hosts = df['host_id'].nunique()
print(f"Number of unique host IDs: {total_hosts}")
```

```python
df['duplicate_host'] = df.duplicated(subset=['host_id'], keep=False)
```

```python
num_duplicates = df['duplicate_host'].sum()
print(f"Number of listings with duplicate host IDs: {num_duplicates}")
```

```python
# How many listings per host
listings_per_host = df.groupby('host_id')['id'].count().sort_values(ascending=False)
print("Top 10 hosts by number of listings:\n", listings_per_host.head(10))
```

```python
# Count how many hosts have 1,2,3,4,5,6+ listings
def listing_bin(x):
    if x >= 10:
        return '6+ listings'
    else:
        return f'{x} listing' if x==1 else f'{x} listings'

# Apply binning
listings_per_host_bins = listings_per_host.apply(listing_bin)

# Count number of hosts in each bin
hosts_count_by_bin = listings_per_host_bins.value_counts().sort_index()

# --- Plot ---
pastel_colors = ['#779ECB', '#FF6961', '#77DD77', '#FFB347', '#B39EB5', '#FDFD96']

fig = px.bar(
    x=hosts_count_by_bin.index,
    y=hosts_count_by_bin.values,
    color=hosts_count_by_bin.index,
    color_discrete_sequence=pastel_colors,
    text=hosts_count_by_bin.values,
    title="Number of Hosts by Number of Listings",
    labels={'x':'Number of Listings per Host', 'y':'Number of Hosts'}
)

fig.update_traces(textposition='outside')
fig.update_layout(showlegend=False)
fig.show()
```

```python
duplicate_hosts = df[df['duplicate_host']].groupby('host_id')['name'].count().sort_values(ascending=False)
print("Duplicate hosts (host_id -> number of listings):\n", duplicate_hosts)
```

```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Map: highlight duplicate hosts
fig_map = px.scatter_mapbox(
    df,
    lat='latitude',
    lon='longitude',
    color='duplicate_host',          # True = duplicate host
    hover_name='host_name',
    hover_data=['host_id', 'name', 'price'],
    color_discrete_map={True:'green', False:'blue'},
    size='reviews_per_month',        # now safe, NaNs replaced
    size_max=15,
    zoom=11
)

fig_map.update_layout(
    mapbox_style="open-street-map",
    title="Airbnb Listings in Melbourne: Duplicate Hosts Highlighted"
)

fig_map.show()
```

```python
# Hosts with more than 1 listing
duplicate_hosts = listings_per_host[listings_per_host > 1].index
# Filter dataset for properties owned by these hosts
duplicate_host_properties = df[df['host_id'].isin(duplicate_hosts)]

# Count number of duplicate hosts per neighbourhood
# (use host_id to avoid counting multiple listings from same host in same area multiple times)
hosts_per_neighbourhood = duplicate_host_properties.groupby('neighbourhood')['host_id'].nunique().sort_values(ascending=False)

# Convert to DataFrame for plotting
hosts_per_neighbourhood_df = hosts_per_neighbourhood.reset_index()
hosts_per_neighbourhood_df.columns = ['neighbourhood', 'num_duplicate_hosts']

# Plot: Neighbourhood vs Number of Duplicate Hosts
fig = px.bar(
    hosts_per_neighbourhood_df,
    x='neighbourhood',
    y='num_duplicate_hosts',
    text='num_duplicate_hosts',
    title="Neighbourhoods with Most Hosts Owning Multiple Airbnbs",
    labels={'neighbourhood':'Neighbourhood', 'num_duplicate_hosts':'Number of Hosts with >1 Listing'},
    color='num_duplicate_hosts',
    color_continuous_scale=pastel_colors
)

fig.update_traces(textposition='outside')
fig.update_layout(template='plotly_white')
fig.show()
```

```python
import spacy
import textstat
import nltk
import re
from nltk.tokenize import word_tokenize

nltk.download('punkt', download_dir='/usr/share/nltk_data')
nltk.download('punkt_tab')

nlp = spacy.load("en_core_web_sm")
```

```python
reviews['clean_comments'] = reviews['comments'].astype(str).str.lower()
reviews['clean_comments'] = reviews['clean_comments'].apply(lambda x: re.sub(r'[^a-z\s]', '', x))
reviews
```

```python
reviews['tokens'] = reviews['clean_comments'].apply(word_tokenize)
```

```python
len(reviews)
```

```python
reviews['word_count'] = reviews['tokens'].apply(len)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure()
sns.histplot(
    reviews['word_count'],
    bins=80,
    color="#A3C1AD"
)
plt.xlim(0,300)
plt.title('Distribution of Review Length')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.show()
```

```python
import textstat

reviews['readability'] = reviews['clean_comments'].apply(lambda x: textstat.flesch_reading_ease(x))

plt.figure()
sns.histplot(reviews['readability'], bins=50,color="#A3C1AD")
plt.title('Readability Score Distribution')
plt.xlabel('Flesch Reading Ease')
plt.ylabel('Frequency')
plt.show()
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

texts = reviews['clean_comments'].tolist()

# Process in batches
docs = nlp.pipe(texts, batch_size=50, n_process=4)  # n_process>1 for multicore

# Extract POS tags
reviews['pos_tags'] = [[token.pos_ for token in doc] for doc in docs]
```

```python
from collections import Counter

all_pos = [pos for sublist in reviews['pos_tags'] for pos in sublist]
pos_counts = Counter(all_pos)

plt.figure()
sns.barplot(x=list(pos_counts.keys()), y=list(pos_counts.values()))
plt.title('POS Tag Frequency')
plt.xticks(rotation=45)
plt.show()
```

```python
def extract_adj(text):
    doc = nlp(text)
    return [token.lemma_ for token in doc if token.pos_ == 'ADJ']

reviews['adjectives'] = reviews['clean_comments'].apply(extract_adj)
```

```python

adj_words = [word for sublist in reviews['adjectives'] for word in sublist]
adj_counts = Counter(adj_words).most_common(20)

adj_df = pd.DataFrame(adj_counts, columns=['Adjective','Frequency'])

plt.figure()
sns.barplot(data=adj_df, x='Frequency', y='Adjective')
plt.title('Top 20 Adjectives Used in Reviews')
plt.show()
```

```python
green_words = ['park','garden','tree','green','nature','quiet','lake','river','view']

def green_count(tokens):
    return sum([1 for word in tokens if word in green_words])

reviews['green_score'] = reviews['tokens'].apply(green_count)

plt.figure()
sns.histplot(reviews['green_score'], bins=20)
plt.title('Green Space Mention Frequency')
plt.xlabel('Green Score')
plt.ylabel('Frequency')
plt.show()
```

```python
from wordcloud import WordCloud

text = ' '.join(adj_words)

wordcloud = WordCloud(width=800, height=400).generate(text)

plt.figure()
plt.imshow(wordcloud)
plt.axis('off')
plt.title('Adjective WordCloud')
plt.show()
```

```python
def lemmatize(text):
    doc = nlp(text)
    return [token.lemma_ for token in doc]

reviews['lemmas'] = reviews['clean_comments'].apply(lemmatize)
```

```python
def extract_verbs(text):
    doc = nlp(text)
    return [token.lemma_ for token in doc if token.pos_ == 'VERB']

reviews['verbs'] = reviews['clean_comments'].apply(extract_verbs)

verb_words = [word for sublist in reviews['verbs'] for word in sublist]
verb_counts = Counter(verb_words).most_common(20)

verb_df = pd.DataFrame(verb_counts, columns=['Verb','Frequency'])

plt.figure()
sns.barplot(data=verb_df, x='Frequency', y='Verb')
plt.title('Top 20 Verbs Used in Reviews')
plt.show()
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
reviews['sentiment'] = reviews['clean_comments'].apply(lambda x: sia.polarity_scores(x)['compound'])

plt.figure()
sns.histplot(reviews['sentiment'], bins=50)
plt.title('Sentiment Distribution of Reviews')
plt.show()
```

```python
sns.scatterplot(x=reviews['green_score'], y=reviews['sentiment'])
plt.title('Green Mentions vs Sentiment')
plt.show()
```
